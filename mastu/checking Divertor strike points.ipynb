```python
import numpy as np
import matplotlib.pyplot as plt
import my_funcs as mf

%matplotlib widget
```

```python
Rlim, zlim = mf.getLimiter()
```

```python
from pyEquilibrium.equilibrium import equilibrium as equil
```

```python
shot1 = 48347
t = 0.355
eq = equil(device='MASTU', shot=shot1, time=t)
surface = eq.get_fluxsurface(1.0)
R1 = surface.R
Z1 = surface.Z
```

```python
shot2 = 50364
t = 0.520
eq = equil(device='MASTU', shot=shot2, time=t)
surface = eq.get_fluxsurface(1.0)
R2 = surface.R
Z2 = surface.Z
```

```python
shot3 = 50365
t = 0.525
eq = equil(device='MASTU', shot=shot3, time=t)
surface = eq.get_fluxsurface(1.0)
R3 = surface.R
Z3 = surface.Z
```

```python
shot4 = 50478
t = 0.575
eq = equil(device='MASTU', shot=shot4, time=t)
surface = eq.get_fluxsurface(1.0)
R4 = surface.R
Z4 = surface.Z
```

```python
shot5 = 50561
t = 0.340
eq = equil(device='MASTU', shot=shot5, time=t)
surface = eq.get_fluxsurface(1.0)
R5 = surface.R
Z5 = surface.Z
```

```python
fig, ax = plt.subplots(1, 2, figsize=(6.5,4.5), dpi=150)
ax[0].plot(Rlim, zlim, '-k')
ax[0].set_aspect('equal')
ax[0].set_xlabel('$R$ (m)', fontsize=9)
ax[0].set_ylabel('$z$ (m)', fontsize=9)
ax[0].tick_params(axis="both", which='both', labelsize=9, direction='in', 
                    left=True, bottom=True, right=True, top=True)
ax[0].set_ylim((ax[0].get_ylim()[0],-0.4))
ax[0].plot(R1, Z1, c='C0', label=shot1)
ax[0].plot(R2, Z2, c='C1', label=shot2)
ax[0].plot(R3, Z3, c='C2', label=shot3)
ax[0].plot(R4, Z4, c='C3', label=shot4)
ax[0].plot(R5, Z5, c='C4', label=shot5)
ax[0].legend(fancybox=1, framealpha=1, fontsize=8, loc='center right')

ax[1].plot(Rlim, zlim, '-k')
ax[1].set_aspect('equal')
ax[1].set_xlabel('$R$ (m)', fontsize=9)
ax[1].set_ylabel('$z$ (m)', fontsize=9)
ax[1].tick_params(axis="both", which='both', labelsize=9, direction='in', 
                    left=True, bottom=True, right=True, top=True)
ax[1].set_ylim((0.4,ax[1].get_ylim()[1]))
ax[1].plot(R1, Z1, c='C0', label=shot1)
ax[1].plot(R2, Z2, c='C1', label=shot2)
ax[1].plot(R3, Z3, c='C2', label=shot3)
ax[1].plot(R4, Z4, c='C3', label=shot4)
ax[1].plot(R5, Z5, c='C4', label=shot5)
ax[1].legend(fancybox=1, framealpha=1, fontsize=8, loc='center right')

plt.tight_layout()
```
