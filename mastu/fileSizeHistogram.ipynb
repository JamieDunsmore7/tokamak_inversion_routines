```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib widget
```

```python
data = np.loadtxt('rba_filesize.txt')
```

```python
size0 = 1.

hist, be = np.histogram(data[:,1]/size0, bins=np.arange(25., 95.1, 10.)/size0)

fig, ax = plt.subplots(1, 1, figsize=(3.5,3), dpi=150)
ax.bar(be[:-1]+np.diff(be).mean(), hist, width=np.diff(be).mean(), edgecolor='k')
if np.isclose(size0, 1.):
    ax.set_xlabel(f'Filesize (MB)', fontsize=9)
else:
    ax.set_xlabel(f'Filesize/{size0:.0f}', fontsize=9)
ax.set_ylabel('Count', fontsize=9)
ax.tick_params(axis="both", which='both', labelsize=9, direction='in', 
                    left=True, bottom=True, right=True, top=True)
ax.set_xticks(np.arange(30., 100.1, 10.)/size0)
ax.set_title(f'# files: {data.shape[0]}')
plt.tight_layout()
```
