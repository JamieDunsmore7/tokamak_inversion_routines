```python
import functions as fn
import numpy as np
from mastu import HSV
import matplotlib.pyplot as plt
import my_funcs as mf
from pyEquilibrium.equilibrium import equilibrium as equil
import pyuda
client = pyuda.Client()
from scipy.optimize import curve_fit

%matplotlib widget
```

```python
from importlib import reload
import quickplot as qp
```

```python
shotn = 51197

trange = [0.0,0.7]

Rwant = [0.98,1.18,1.38]
zwant = 0.

alpha = 0.3
dp = 3
fontsize = 9
```

```python
### get the thomson data and chop it down
tth, Te, TeErr, Rth = qp.getData(shotn, 'T_e', trange=trange, Rwant=Rwant)
_, ne, neErr, _ = qp.getData(shotn, 'n_e', trange=trange, Rwant=Rwant)

### calculate the collisionality
coulombLog = 24. - np.log(np.sqrt(ne / 1e6) / Te)
collision = 2.91e-6 * (ne / 1e6) * coulombLog / (Te**1.5)
```

```python
### plot time trace for Te
figTe, axTe = qp.plotTe(
    shotn, tth, Te, TeErr, Rth, alpha=0.3, dp=3, fontsize=fontsize
)
### plot time trace for ne
figne, axne = qp.plotne(
    shotn, tth, ne, neErr, Rth, alpha=0.3, dp=3, fontsize=fontsize
)
### plot time trace for collisionality
fignu, axnu = qp.plotCollision(
    shotn, tth, collision, Rth, dp=dp, fontsize=fontsize
)
```

```python
### get magnetic field
Bt, Btot = qp.getB(shotn, tth, Rwant=Rwant, zwant=zwant)
```

```python
### plot time trace for Te/Bt
figTeBt, axTeBt = qp.plotTeBt(
    shotn, tth, Te, Rth, Bt, Btot, dp=dp, fontsize=fontsize
)
```

```python
### calculate electron gyroradius
rhoE = mf.gyroradius(Te, Btot, m=1., particle='e')

### get minor radius
minorRadius, minorTime = mf.getMinorRadius(shotn, trange=trange)
### make rhoStar
rhoStarE = rhoE / np.interp(tth, minorTime, minorRadius)[:,np.newaxis]
```

```python
### get Ti
try:
    tcx, Ti, TiErr, Rcx = qp.getData(shotn, 'T_i', trange=trange, Rwant=Rwant)
    ### calculate ion gyroradius
    rhoI = np.zeros_like(Ti)
    for i in range(0, 3):
        rhoI[:,i] = mf.gyroradius(
            Ti[:,i], np.interp(tcx, tth, Btot[:,i]), m=2., particle='i'
        )
    rhoStarI = rhoI / np.interp(tcx, minorTime, minorRadius)[:,np.newaxis]
    print('success')
except:
    tcx = None
    Ti = None
    TiErr = None
    Rcx = None
    rhoI = None
    rhoStarI = None
    print('failed')
```

```python
### plot rhoStar
figRho, axRho = qp.plotRhoStar(
    shotn, tth, rhoStarE, Rth, rhoI=rhoStarI, tI=tcx, dp=dp, fontsize=fontsize
)
```

```python
twant = 0.1

Rplot = np.arange(0.3, 1.5, 0.001)
eq = equil(device='MASTU', shot=shotn, time=twant)
Btor = np.abs(eq.Bt(Rplot, 0.)[0])
Bphi = np.abs(eq.Bphi(Rplot, 0.)[0])
Btotal = eq.B(Rplot, 0.)[0]
Bpol = eq.Bp(Rplot, 0.)[0]
BR = eq.BR(Rplot, 0.)[0]

fig, ax = plt.subplots(1, 1, figsize=(3.5,3), dpi=150)
ax.plot(Rplot, Btotal, label='$B_\\mathrm{tot}$')
ax.plot(Rplot, Btor, label='$B_t$')
ax.plot(Rplot, Bphi, label='$B_\\phi$')
ax.plot(Rplot, Bpol, label='$B_\\theta$')
ax.plot(Rplot, BR, label='$B_R$')
ax.set_xlabel('$R$ - (m)', fontsize=fontsize)
ax.set_ylabel('$B$ - (T)', fontsize=fontsize)
ax.legend(
    fancybox=1, framealpha=1, fontsize=fontsize-1,  
    handlelength=0, handletextpad=0, labelcolor='linecolor'
)
ax.tick_params(
    axis="both", which='both', labelsize=8, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax.minorticks_on()
ax.set_title(f'shot # {shotn}, $t$ = {twant:.3f}s', fontsize=fontsize)
plt.tight_layout()
```
