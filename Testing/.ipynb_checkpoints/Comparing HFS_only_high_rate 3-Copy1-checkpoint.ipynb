```python
def rolling(arr, w):
    return np.convolve(arr, np.ones(w), mode='same') / w

def rollingMedian(arr, w):
    result = np.zeros_like(arr)
    I = w // 2
    result[:I] = np.nan
    result[-I:] = np.nan
    for i in range(I, len(arr)-I):
        result[i] = np.median(arr[i-I:i+I])
    return result
```

```python
shotns = [
    51691, 51704, 51705, 51706, 51709, 51710, 51711, 51712, 51713, 51761, 51762, 51763, 51764, 51765
]
# boo =    [
#      True,  True, False, False, False, False, False, False, False,  True, False, False, False, False
# ]
boo =    [
    False,  True, False, False, False, False, False, False, False, False, False, False, False, False
]
colors = [ 
      'k',  'C0',  'C1',  'C2',  'C3',  'C4',  'C5',  'C6',  'C7',  'C8',  'C9', 'C10', 'C11', 'C12'
]

fuels = [
    '/xdc/flow/s/hfs_mid_u02', 
#     '/xdc/flow/s/lfsv_bot_l03', 
#     '/xdc/flow/s/lfsv_top_u11', 
#     '/xdc/flow/s/lfsd_bot_l0506', 
#     '/xdc/flow/s/lfsd_top_u0506', 
#     '/xdc/flow/s/pfr_bot_b01', 
#     '/xdc/flow/s/pfr_top_t05', 
#     '/xdc/flow/s/hfs_mid_l02', 
]

puffs = [0.23, 0.71]
# puffs = [0.23, 0.35, 0.47, 0.59, 0.71]

time = [None] * len(shotns)

emMax = [None] * len(shotns)
emMaxR = [None] * len(shotns)
emMaxErr = [None] * len(shotns)

n0Max = [None] * len(shotns)
n0MaxR = [None] * len(shotns)
n0MaxErr = [None] * len(shotns)

n0Max2 = [None] * len(shotns)
n0Max2R = [None] * len(shotns)
n0Max2Err = [None] * len(shotns)
n0boo = [None] * len(shotns)

n0Sep = [None] * len(shotns)
n0SepErr = [None] * len(shotns)

sizMax = [None] * len(shotns)
sizMaxR = [None] * len(shotns)
sizMaxErr = [None] * len(shotns)

sizMax2 = [None] * len(shotns)
sizMax2R = [None] * len(shotns)
sizMax2Err = [None] * len(shotns)
sizboo = [None] * len(shotns)

sepR = [None] * len(shotns)

for i, shotn in enumerate(shotns):
    if boo[i]:
        if shotn != 51762:
            
            resultsFile = f'/home/sthoma/Documents/Results/rba/{shotn}/rba_results_neutral_1_A.npz'
            results = np.load(resultsFile)

            time[i] = results['time']
            
            emMax[i] = results['emMax']
            emMaxR[i] = results['emMaxR']
                        
            n0Max[i] = results['neutralMax']
            n0MaxR[i] = results['neutralMaxR']
            n0MaxErr[i] = results['neutralMaxErr']
            
            n0Max2[i] = results['neutralMaxInside']
            n0Max2Err[i] = results['neutralMaxInsideErr']
            n0Max2R[i] = results['neutralMaxInsideR']
            n0boo[i] = results['neutralBoo']
            
            sizMax[i] = results['ioniseMax']
            sizMaxR[i] = results['ioniseMaxR']
            sizMaxErr[i] = results['ioniseMaxErr']
            
            sizMax2[i] = results['ioniseMaxInside']
            sizMax2Err[i] = results['ioniseMaxInsideErr']
            sizMax2R[i] = results['ioniseMaxInsideR']
            sizboo[i] = results['ioniseBoo']
            
            
            ###########################################################
            R = results['RgridB']
            err = results['emissivityErr']
            
            emMaxErr[i] = np.zeros_like(emMax[i])
            
            
            psiN = results['psiN']
            n0 = results['neutralDensity']
            n0Err = results['neutralErr']
            
            n0Sep[i] = np.zeros_like(n0Max[i])
            n0SepErr[i] = np.zeros_like(n0Max[i])
            
            sepR[i] = np.zeros_like(n0Max[i])
            
            for j in range(0, err.shape[0]):
                
                emMaxErr[i][j] = err[j,fn.findNearest(R, emMaxR[i][j])]
                
                sepInd = fn.findNearest(psiN[j], 1.)
                sepR[i][j] = R[sepInd]
                n0Sep[i][j] = n0[j,sepInd]
                n0SepErr[i][j] = n0Err[j,sepInd]
```
