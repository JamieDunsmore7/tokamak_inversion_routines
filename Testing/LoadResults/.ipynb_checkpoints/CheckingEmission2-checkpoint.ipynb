```python
cmap=plt.cm.viridis

fig, ax = plt.subplots(1, 2, figsize=(6,2.75), dpi=150)


N = 15
levels = np.linspace(0., 3.e-15*(10**N), 13)
cf = ax[0].contour(
    Te0[jslice], ne0[islice], dataExcite[islice,jslice]*(10**N), 
    levels, cmap=cmap, linewidths=1.
)
cbar=fig.colorbar(
    cf, ax=ax[0], ticks=levels[::2], 
)
cbar.ax.tick_params(labelsize=9)
cbar.set_label(
    f'($\\times10^{{{-N:.0f}}}$ ph m$^{{-3}}$ s$^{{-1}}$)', fontsize=9
)
ax[0].set_xscale('log')
ax[0].set_yscale('log')
ax[0].clabel(
    cf, inline=1, fontsize=8, fmt='%1.2f', inline_spacing=5, use_clabeltext=False
)
ax[0].set_xlabel('$T_e$ (eV)', fontsize=9)
ax[0].set_ylabel('$n_e$ (m$^3$)', fontsize=9)
ax[0].tick_params(
    axis="both", which='both', labelsize=9, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax[0].set_title('regular', fontsize=9)
ax[0].set_xlim([Te0[jslice].min(), Te0[jslice].max()])
ax[0].set_ylim([ne0[islice].min()*1.1, ne0[islice].max()])


levels = np.linspace(-20, -14, 13)
cf = ax[1].contour(
    Te0[jslice], ne0[islice], np.log10(dataExcite[islice,jslice]), 
    levels, cmap=cmap, linewidths=1.
)
cbar=fig.colorbar(
    cf, ax=ax[1], ticks=levels[::2], 
)
cbar.ax.tick_params(labelsize=9)
cbar.set_label(
    '$\\log_{10}$ (ph m$^3$ s$^{-1}$)', fontsize=9
)
ax[1].set_xscale('log')
ax[1].set_yscale('log')
ax[1].clabel(
    cf, inline=1, fontsize=8, fmt='%1.1f', inline_spacing=10, use_clabeltext=False
)
ax[1].set_xlabel('$T_e$ (eV)', fontsize=9)
ax[1].set_ylabel('$n_e$ (m$^{-3}$)', fontsize=9)
ax[1].tick_params(
    axis="both", which='both', labelsize=9, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax[1].set_title('$\\log_{10}$', fontsize=9)
ax[1].set_xlim([Te0[jslice].min(), Te0[jslice].max()])
ax[1].set_ylim([ne0[islice].min()*1.1, ne0[islice].max()])

fig.suptitle('pec12#h_pju#h0: PEC$^\\mathrm{EXC}_{2 \\rightarrow 1}$ - excitation', fontsize=9)
plt.tight_layout()
```
