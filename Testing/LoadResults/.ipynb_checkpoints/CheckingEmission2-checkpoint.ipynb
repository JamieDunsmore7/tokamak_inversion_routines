```python
from scipy.interpolate import RegularGridInterpolator
```

```python
scale = 10**int(np.log10(np.median(dataIonise)))
scale1 = 10**int(np.log10(np.median(dataIonise)))

fIonise = RegularGridInterpolator(
    (Te0, ne0), dataIonise.T / scale, method='cubic', 
    bounds_error=False, fill_value=None
)
fIonise1 = RegularGridInterpolator(
    (Te1, ne1), dataIonise1.T / scale, method='cubic', 
    bounds_error=False, fill_value=None
)
```

```python
ION = fIonise((TeProfile, neProfile)) * scale
ION1 = fIonise1((TeProfile, neProfile)) * scale1
```

```python
fig, ax = plt.subplots(1, 1, figsize=(3.5,3), dpi=150)

ax.plot(Rprofile, ION, '-', lw=0.8, c='C0', mfc='None', label='ADAS')
ax.plot(Rprofile, ION1, '-', lw=0.8, c='C1', mfc='None', label='JHS')

legend = ax.legend(
    fancybox=1, framealpha=1, handlelength=0, loc='upper right', 
    handletextpad=0, labelcolor='linecolor', fontsize=8
)
for handle in legend.legendHandles:
    handle.set_marker(None)
    
ax.tick_params(
    axis="both", which='both', labelsize=9, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax.minorticks_on()
ax.yaxis.get_offset_text().set_size(9)
ax.set_xlabel('$R$ (m)', fontsize=9)
ax.set_ylabel('Iz. (m$^{-3}$ s$^{-1}$)', fontsize=9)
ax.set_title(f'shot # 51706, t=0.300s', fontsize=9)

plt.tight_layout()
```
