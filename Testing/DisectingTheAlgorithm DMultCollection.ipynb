```python
import matplotlib.pyplot as plt
%matplotlib widget
```

```python
shotn = 50499
DMults = [1., 2., 4., 8., 16., 32., 64., 128., 512., 1024., 5000., 10000.]


xlim = [1.,1.90]


fig, ax = plt.subplots(1, 2, figsize=(5.5,3), dpi=150)

for i, DMult in enumerate(DMults):
    d = np.load(f'/home/sthoma/tokamak_inversion_routines/Testing/TestingAlgorithm/DMult/{shotn}_Dmult_{DMult:.4f}.npy.npz')
    R = d['R']
    RgridB = d['RgridB']
    Y = d['y']
    yErr = d['yErr']
    data = d['data']
    err = d['err']
    backprojection = d['backprojection']
    
    if i == 0:
        ax[0].plot(R, data, c='k', label='HSV', zorder=0)
    ax[0].fill_between(
        R, data-err, data+err, color='k', alpha=0.1, zorder=0
    )
        
    ax[0].plot(R, backprojection, c=f'C{i}', label=f'{DMult:.0f}')
    
    ax[1].plot(RgridB, Y, c=f'C{i}', label=f'{DMult:.0f}')
    ax[1].fill_between(
        RgridB, Y-yErr, Y+yErr, color=f'C{i}', alpha=0.3
    )


for i in range(0, 2):
    ax[i].set_xlim(xlim)
    ax[i].plot(xlim, [0.,0.], '-k', lw=0.8, zorder=1)
    ax[i].legend(
        fancybox=1, framealpha=1, fontsize=8, ncol=2, 
        handlelength=0, handletextpad=0, labelcolor='linecolor'
    )
    ax[i].yaxis.get_offset_text().set_size(9)
    ax[i].tick_params(
        axis="both", which='both', labelsize=9, direction='in', 
        left=True, bottom=True, right=True, top=True
    )
    ax[i].minorticks_on()

ax[0].set_ylim([-0.01e20, 0.3e20])
ax[1].set_ylim([-0.05e20, 0.5e20])

ax[0].set_xlabel('R (m)', fontsize=9)
ax[0].set_ylabel('Brightness ($\\gamma$/m$^2$/sr/s)', fontsize=9)
ax[1].set_xlabel('R (m)', fontsize=9)
ax[1].set_ylabel('Emissivity - $\\epsilon$ ($\\gamma$/m$^3$/sr/s)', fontsize=9)
plt.suptitle(f'shot # {shotn}', fontsize=9)

plt.tight_layout()
```
