```python
yplot0 = data0
yplot1 = data1
yplot2 = data2
ylabel = 'Brightness'

fig, ax = plt.subplots(2, 1, figsize=(3.5,4.0), dpi=150, sharex=True)
ax[0].plot(R0, yplot0, label='D0')
ax[0].plot(R1, yplot1, label='D1')
ax[0].plot(R2, yplot2, label='D2')
xlim = ax[0].get_xlim()
ax[0].set_xlim(xlim)
ax[0].plot(xlim, [0.,0.], '-k', lw=0.8, zorder=0)
ax[0].legend(
    fancybox=1, framealpha=1, fontsize=8, 
    handlelength=0, handletextpad=0, labelcolor='linecolor'
)
ax[1].plot(R0, yplot0-yplot1)
ax[1].plot(R0, yplot1-yplot2)
xlim = ax[1].get_xlim()
ax[1].set_xlim(xlim)
ax[1].plot(xlim, [0.,0.], '-k', lw=0.8, zorder=0)

ax[0].set_ylabel(ylabel, fontsize=9)
ax[1].set_xlabel('R (m)', fontsize=9)
ax[1].set_ylabel(ylabel, fontsize=9)

ax[0].tick_params(
    axis="both", which='both', labelsize=9, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax[0].minorticks_on()
ax[0].yaxis.get_offset_text().set_size(9)
ax[1].tick_params(
    axis="both", which='both', labelsize=9, direction='in', 
    left=True, bottom=True, right=True, top=True
)
ax[1].minorticks_on()
ax[1].yaxis.get_offset_text().set_size(9)

plt.tight_layout()
```
