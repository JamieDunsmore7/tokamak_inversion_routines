```python
data = np.load('/home/sthoma/Documents/Results/MU04/Inversions/sys_err_5_scale_2.5x.npz')
data.files
R = data['R']
frame = data['frame']
R_grid = data['R_grid']
y = data['y']
y_err = data['y_err']
backprojection = data['backprojection'][0]

n = 20
R = np.zeros((n, len(R)))
frame = np.zeros((n, len(frame)))
R_grid = np.zeros((n, len(R_grid)))
y = np.zeros((n, len(y)))
y_err = np.zeros((n, len(y_err)))
backprojection = np.zeros((n, len(backprojection)))

for i in range(0, 20):
    data = np.load(f'/home/sthoma/Documents/Results/MU04/Inversions/sys_err_{i+1}_scale_2.5x.npz')
    R[i] = data['R']
    frame[i] = data['frame']
    R_grid[i] = data['R_grid']
    y[i] = data['y']
    y_err[i] = data['y_err']
    backprojection[i] = data['backprojection'] 
```

```python
from matplotlib.colors import Normalize
def normalise(xx):
    vmin = xx.min()
    vmax = xx.max()
    return (xx - vmin)/(vmax-vmin)
```

```python
R_grid_b = (R_grid[0,1:]+ R_grid[0,:-1])/2
```

```python
sys_err = np.arange(1, 20.1, 1)
cmap = plt.cm.viridis
norm = Normalize(vmin=sys_err[0], vmax=sys_err[-1])
colors = normalise(sys_err)
c = cmap(colors)

xplot = [0.2,1.875]

fig, ax = plt.subplots(1, 1, figsize=(3.5,3), dpi=150)
ax.plot(R[0], frame[0], '-', c='k', zorder=3, label='raw RBA')

i = 0
ax.plot(R_grid_b, y[i], '-', c='C0', zorder=2, label=f'sys_err={i+1}')
ax.fill_between(R_grid_b, y[i]-y_err[i], y[i]+y_err[i], color='C0', alpha=0.2, zorder=2)

i = 4
ax.plot(R_grid_b, y[i], '-', c='C1', zorder=2, label=f'sys_err={i+1}')
ax.fill_between(R_grid_b, y[i]-y_err[i], y[i]+y_err[i], color='C1', alpha=0.2, zorder=2)

i = 9
ax.plot(R_grid_b, y[i], '-', c='C2', zorder=2, label=f'sys_err={i+1}')
ax.fill_between(R_grid_b, y[i]-y_err[i], y[i]+y_err[i], color='C2', alpha=0.2, zorder=2)

i = 14
ax.plot(R_grid_b, y[i], '-', c='C3', zorder=2, label=f'sys_err={i+1}')
ax.fill_between(R_grid_b, y[i]-y_err[i], y[i]+y_err[i], color='C3', alpha=0.2, zorder=2)

i = 19
ax.plot(R_grid_b, y[i], '-', c='C4', zorder=2, label=f'sys_err={i+1}')
ax.fill_between(R_grid_b, y[i]-y_err[i], y[i]+y_err[i], color='C4', alpha=0.2, zorder=2)

ax.set_xlim(xplot)
ax.plot(xplot, [0.,0.], '-k', lw=0.8, zorder=1)
ax.set_xlabel('R (m)', fontsize=9)
ax.set_ylabel('Units', fontsize=9)
ax.legend(fancybox=1, framealpha=1, fontsize=8, ncol=2)
ax.tick_params(axis="both", which='both', labelsize=9, direction='in', 
              left=True, bottom=True, right=True, top=False)

plt.tight_layout()
```
